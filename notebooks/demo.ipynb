## 3. Esegui il sistema multi-agente

```python
from src.llm_client import LLMClient
from src.orchestrator import MultiAgentOrchestrator

llm = LLMClient()
print("LLM:", llm)

orch = MultiAgentOrchestrator(llm=llm)

def step(agent, out):
    print(f"\n{'='*60}\n🤖 {agent}\n{'='*60}\n{out[:1500]}")

task = "Spiega cos'e' il cloud computing, i suoi vantaggi e 3 servizi principali."
res = orch.run_pipeline(task, on_step=step)

print(f"\n{'#'*60}\nRISULTATO FINALE:\n{res['final']}\n{'#'*60}")
```

## 4. Modalita' interattiva (chat)

```python
# Chat semplice con memoria
from src.memory import ConversationMemory
mem = ConversationMemory()
while True:
    q = input("Tu> ").strip()
    if not q or q.lower() in ("exit","quit"): break
    mem.add("user", q)
    r = orch.run_pipeline(q)
    mem.add("assistant", r["final"])
    print("\nAgente>", r["final"][:2000])
```
